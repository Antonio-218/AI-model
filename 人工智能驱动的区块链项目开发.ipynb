# 加载模型与分词器

```python
 #加载预训练模型和分词器
from unsloth import FastLanguageModel  # 导入unsloth中的FastLanguageModel类
from transformers import AutoTokenizer  # 导入transformers库中的AutoTokenizer类
import torch  # 导入PyTorch库

# 加载模型和分词器
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/DeepSeek-R1-Distill-Qwen-7B-unsloth-bnb-4bit",  # 指定模型名称
    max_seq_length=2048,  # 设置模型的最大序列长度
    dtype=torch.bfloat16,  # 使用bfloat16数据类型以优化计算效率
    load_in_4bit=True  # 以4位量化方式加载模型以节省内存
)
```

# 注入LORA模块

```python
# 打印加载的模型和分词器信息
print(f"Model: {model}")
print(f"Tokenizer: {tokenizer}")

# 为模型注入LORA模块，以实现高效的模型微调
from peft import LoraConfig, get_peft_model  # 导入peft库中的LoraConfig和get_peft_model函数

# 配置并注入LORA模块
model = FastLanguageModel.get_peft_model(
    model,  # 待修改的模型
    r=64,  # LORA的秩（r），控制参数数量
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],  # 指定需要注入LORA模块的目标模块
    lora_alpha=16,  # LORA的缩放比例
    lora_dropout=0.05,  # LORA模块的dropout概率
    bias="none",  # 不使用偏置项
    use_gradient_checkpointing=True  # 启用梯度检查点以节省内存
)
```

# 微调前验证生成效果

```python
# 在微调之前验证模型的初始生成效果
prompt = "交易哈希为 0x1630239127e29a29c51f0855c5ef1100775901621e9ff483b9c97f5ca3c44e0b 的交易用了多少 gas？"  # 提供一个测试文本作为提示

# 将提示文本转化为模型输入
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)  # 使用分词器对提示文本编码，并将其发送到模型所在设备

# 使用模型生成文本
outputs = model.generate(**inputs, max_new_tokens=300)  # 根据输入生成文本，最多生成300个新token

# 解码并打印生成的文本结果
print(tokenizer.decode(outputs[0], skip_special_tokens=True))  # 将生成的token序列解码为文本，并跳过特殊标记
```

# 加载数据集

```python
from datasets import load_dataset
# 从datasets库中导入了load_dataset函数，用于加载数据集

dataset = load_dataset("json", data_files={"train": "erc20_chat_train_samples.json"}, split="train")
# 从一个JSON文件中加载数据集，并将其指定为训练集

dataset = dataset.train_test_split(test_size=0.1)
# 0.1表示在进行train_test_split操作时，将数据集划分成训练集和测试集，其中测试集的大小占整个数据集的10%
print(dataset["train"][0])
# 打印出加载的训练集中的第4个样本
```

```python
# 定义 formatting_func 用于转成训练样本
def formatting_func(example):
    messages = example["messages"]
    user_message = ""
    assistant_message = ""
    #初始化两个空字符串，用于存储用户和助手的消息
    for message in messages:
        if message["role"] == "user":
            user_message += message["content"].strip() + "\n"
        elif message["role"] == "assistant":
            assistant_message += message["content"].strip() + "\n"

    formatted = f"""### 用户问题：
{user_message.strip()}
#遍历 messages 列表中的每个消息对象，
#根据消息的 role 字段（表示消息是由用户还是助手发送的），将消息内容添加到相应的字符串中。
#strip() 方法用于移除消息内容两端的空白字符，\n 用于在消息之间添加换行符。
### 机器人回答：
{assistant_message.strip()}"""
    return [formatted]  # 必须返回 list
print(formatting_func(dataset["train"][2])[0])
```

# 设置训练参数

```python
from transformers import TrainingArguments

# 初始化 TrainingArguments 对象
training_args = TrainingArguments(
    # 输出目录，用于保存模型和日志文件
    output_dir = "outputs",
    # 每个设备的训练批次大小
    per_device_train_batch_size = 2,
    # 每个设备的评估批次大小
    per_device_eval_batch_size = 2,
    # 梯度累积步数，用于模拟更大批次大小的训练
    gradient_accumulation_steps = 2,
    # 预热步数，用于逐步调整学习率
    warmup_steps = 10,
    # 最大训练步数，达到这个步数后训练停止
    max_steps = 50,
    # 学习率，用于控制模型权重更新的速度
    learning_rate = 2e-4,
    # 日志记录步数，每隔多少步记录一次日志
    logging_steps = 1,
    # 是否启用混合精度训练，使用 float16 提高训练速度和减少显存占用
    fp16 = True,
    # 保存模型的步数间隔
    save_steps = 25,
    # 评估策略，"steps" 表示每隔一定步数进行评估
    eval_strategy = "steps",
    # 评估步数间隔
    eval_steps = 10,
    # 保存模型的总数量上限
    save_total_limit = 1,
    # 报告工具，"none" 表示不使用任何报告工具
    report_to = "none"
)
```

# 格式转换

```python
def apply_format(example):
    return {"text": formatting_func(example)[0]}# 将格式化后的文本存入"text"字段

dataset = dataset.map(apply_format)# 应用格式化函数到整个数据集
print(dataset["train"][0])  # 打印训练集的第1条样本
```

# 训练器初始化

```python
# 从unsloth.trainer模块导入SFTTrainer类
from unsloth.trainer import SFTTrainer

# 初始化SFTTrainer对象
trainer = SFTTrainer(
    # 指定要训练的模型
    model = model,
    # 指定用于数据预处理和编码的tokenizer
    tokenizer = tokenizer,
    # 指定训练数据集
    train_dataset = dataset["train"],
    # 指定评估数据集
    eval_dataset = dataset["test"],
    # 指定数据集中包含文本的字段名称
    dataset_text_field = "input",
    # 指定序列的最大长度，超过这个长度的序列将被截断
    max_seq_length = 2048,
    # 指定训练参数
    args = training_args,
)
```

# 模型训练

```python
# // 开始训练
trainer.train()
```

# 微调后的模型验证

```python
# 定义用户的查询，这个查询是关于一个特定交易哈希的交易使用的gas数量
prompt = "### 用户问题:\n交易哈希为 0x1630239127e29a29c51f0855c5ef1100775901621e9ff483b9c97f5ca3c44e0b 的交易用了多少 gas?"

# 使用tokenizer对输入的文本进行编码，返回PyTorch张量，并指定这些张量应该被转移到模型所在的设备上（如GPU）
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# 调用模型的generate方法，使用编码后的输入来生成文本
# max_new_tokens=300限制了生成文本的最大长度为300个token
outputs = model.generate(**inputs, max_new_tokens=300)

# 使用tokenizer将生成的token序列解码回文本形式，并跳过特殊token（如填充token），使输出更易于阅读
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
